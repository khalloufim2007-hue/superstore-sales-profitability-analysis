```python
# ============================================================
# SUPERSTORE - COMPLETE BUSINESS ANALYSIS
# ============================================================

# ============================================================
# 1. REGION PERFORMANCE
# ============================================================

region_performance = df.groupby('Region').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

region_performance['Profit Margin'] = (
    region_performance['Profit'] /
    region_performance['Sales']
) * 100

region_performance = region_performance.sort_values(
    'Profit',
    ascending=False
)

print("========== REGION PERFORMANCE ==========")
display(region_performance)


# ============================================================
# 2. CUSTOMER SEGMENT PERFORMANCE
# ============================================================

segment_performance = df.groupby('Segment').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Orders=('Order ID', 'nunique'),
    Customers=('Customer ID', 'nunique')
)

segment_performance['Profit Margin'] = (
    segment_performance['Profit'] /
    segment_performance['Sales']
) * 100

segment_performance = segment_performance.sort_values(
    'Profit',
    ascending=False
)

print("\n========== CUSTOMER SEGMENT PERFORMANCE ==========")
display(segment_performance)


# ============================================================
# 3. TOP 10 PRODUCTS BY PROFIT
# ============================================================

top_products = df.groupby(
    ['Product ID', 'Product Name']
).agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
).sort_values(
    'Profit',
    ascending=False
).head(10)

print("\n========== TOP 10 PRODUCTS BY PROFIT ==========")
display(top_products)


# ============================================================
# 4. TOP 10 LOSS-MAKING PRODUCTS
# ============================================================

loss_products = df.groupby(
    ['Product ID', 'Product Name']
).agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
).sort_values(
    'Profit',
    ascending=True
).head(10)

print("\n========== TOP 10 LOSS-MAKING PRODUCTS ==========")
display(loss_products)


# ============================================================
# 5. DISCOUNT VS PROFIT
# ============================================================

discount_analysis = df.groupby('Discount').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

discount_analysis['Profit Margin'] = (
    discount_analysis['Profit'] /
    discount_analysis['Sales']
) * 100

print("\n========== DISCOUNT VS PROFIT ==========")
display(discount_analysis)


# ============================================================
# 6. YEARLY PERFORMANCE
# ============================================================

df['Year'] = df['Order Date'].dt.year

yearly_performance = df.groupby('Year').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Orders=('Order ID', 'nunique')
)

yearly_performance['Profit Margin'] = (
    yearly_performance['Profit'] /
    yearly_performance['Sales']
) * 100

print("\n========== YEARLY PERFORMANCE ==========")
display(yearly_performance)


# ============================================================
# 7. MONTHLY PERFORMANCE
# ============================================================

df['Month'] = df['Order Date'].dt.to_period('M')

monthly_performance = df.groupby('Month').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

print("\n========== MONTHLY PERFORMANCE ==========")
display(monthly_performance)


# ============================================================
# 8. CATEGORY + REGION
# ============================================================

category_region = df.groupby(
    ['Region', 'Category']
).agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

category_region['Profit Margin'] = (
    category_region['Profit'] /
    category_region['Sales']
) * 100

print("\n========== REGION + CATEGORY ==========")
display(category_region)


# ============================================================
# 9. SEGMENT + CATEGORY
# ============================================================

segment_category = df.groupby(
    ['Segment', 'Category']
).agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

segment_category['Profit Margin'] = (
    segment_category['Profit'] /
    segment_category['Sales']
) * 100

print("\n========== SEGMENT + CATEGORY ==========")
display(segment_category)


# ============================================================
# 10. TOP CUSTOMERS BY PROFIT
# ============================================================

top_customers = df.groupby(
    ['Customer ID', 'Customer Name']
).agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Orders=('Order ID', 'nunique')
).sort_values(
    'Profit',
    ascending=False
).head(10)

print("\n========== TOP 10 CUSTOMERS BY PROFIT ==========")
display(top_customers)


# ============================================================
# 11. LOSS-MAKING SUB-CATEGORIES
# ============================================================

loss_subcategories = subcategory_performance[
    subcategory_performance['Profit'] < 0
].sort_values(
    'Profit'
)

print("\n========== LOSS-MAKING SUB-CATEGORIES ==========")
display(loss_subcategories)


# ============================================================
# 12. ORDER ECONOMICS
# ============================================================

total_orders = df['Order ID'].nunique()
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()

average_sales_order = total_sales / total_orders
average_profit_order = total_profit / total_orders

print("\n========== ORDER ECONOMICS ==========")
print(f"Total Orders: {total_orders:,}")
print(f"Average Sales per Order: ${average_sales_order:,.2f}")
print(f"Average Profit per Order: ${average_profit_order:,.2f}")


print("\n================================================")
print("        COMPLETE BUSINESS ANALYSIS DONE")
print("================================================")
```

```python
# ============================================================
# SUPERSTORE - ALL PYTHON VISUALIZATIONS
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")


# ============================================================
# 1. SALES BY CATEGORY
# ============================================================

category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=category_sales.index, y=category_sales.values)
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales ($)')
plt.tight_layout()
plt.show()


# ============================================================
# 2. PROFIT BY CATEGORY
# ============================================================

category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=category_profit.index, y=category_profit.values)
plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit ($)')
plt.tight_layout()
plt.show()


# ============================================================
# 3. SALES AND PROFIT BY YEAR
# ============================================================

year_data = df.groupby(
    df['Order Date'].dt.year
).agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

plt.figure(figsize=(9,5))
plt.plot(year_data.index, year_data['Sales'], marker='o', label='Sales')
plt.plot(year_data.index, year_data['Profit'], marker='o', label='Profit')

plt.title('Sales and Profit by Year')
plt.xlabel('Year')
plt.ylabel('Amount ($)')
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 4. PROFIT MARGIN BY CATEGORY
# ============================================================

category_summary = df.groupby('Category').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

category_summary['Profit Margin'] = (
    category_summary['Profit'] /
    category_summary['Sales']
) * 100

category_margin = category_summary['Profit Margin'].sort_values(
    ascending=False
)

plt.figure(figsize=(8,5))
sns.barplot(x=category_margin.index, y=category_margin.values)
plt.title('Profit Margin by Category')
plt.xlabel('Category')
plt.ylabel('Profit Margin (%)')
plt.tight_layout()
plt.show()


# ============================================================
# 5. SALES BY REGION
# ============================================================

region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=region_sales.index, y=region_sales.values)
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales ($)')
plt.tight_layout()
plt.show()


# ============================================================
# 6. PROFIT BY REGION
# ============================================================

region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=region_profit.index, y=region_profit.values)
plt.title('Profit by Region')
plt.xlabel('Region')
plt.ylabel('Profit ($)')
plt.tight_layout()
plt.show()


# ============================================================
# 7. SALES BY CUSTOMER SEGMENT
# ============================================================

segment_sales = df.groupby('Segment')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=segment_sales.index, y=segment_sales.values)
plt.title('Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Sales ($)')
plt.tight_layout()
plt.show()


# ============================================================
# 8. PROFIT BY CUSTOMER SEGMENT
# ============================================================

segment_profit = df.groupby('Segment')['Profit'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=segment_profit.index, y=segment_profit.values)
plt.title('Profit by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Profit ($)')
plt.tight_layout()
plt.show()


# ============================================================
# 9. TOP 10 PRODUCTS BY PROFIT
# ============================================================

top_products = df.groupby('Product Name')['Profit'].sum().sort_values(
    ascending=False
).head(10).sort_values()

plt.figure(figsize=(10,6))
sns.barplot(x=top_products.values, y=top_products.index)
plt.title('Top 10 Products by Profit')
plt.xlabel('Profit ($)')
plt.ylabel('Product')
plt.tight_layout()
plt.show()


# ============================================================
# 10. TOP 10 LOSS-MAKING PRODUCTS
# ============================================================

loss_products = df.groupby('Product Name')['Profit'].sum().sort_values(
    ascending=True
).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=loss_products.values, y=loss_products.index)
plt.title('Top 10 Loss-Making Products')
plt.xlabel('Profit ($)')
plt.ylabel('Product')
plt.tight_layout()
plt.show()


# ============================================================
# 11. DISCOUNT VS PROFIT
# ============================================================

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='Discount',
    y='Profit',
    alpha=0.5
)

plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit ($)')
plt.tight_layout()
plt.show()


# ============================================================
# 12. PROFIT BY SUB-CATEGORY
# ============================================================

subcategory_profit = df.groupby(
    'Sub-Category'
)['Profit'].sum().sort_values()

plt.figure(figsize=(10,7))
sns.barplot(
    x=subcategory_profit.values,
    y=subcategory_profit.index
)

plt.title('Profit by Sub-Category')
plt.xlabel('Profit ($)')
plt.ylabel('Sub-Category')
plt.tight_layout()
plt.show()


# ============================================================
# COMPLETE
# ============================================================

print("==========================================")
print("ALL VISUALIZATIONS COMPLETED")
print("==========================================")
```
